## A. WORD ROTATION PROBLEM

```python
def rotate_word(word):
    x = len(word)

    return word[1:x:1]+word[0]
```

```python
rotate_word("python")
```

```python
rotate_word("logic")
```

```python
rotate_word("Code")
```

```python
rotate_word("A")
```

## B. USERNAME BUILDER PROBLEM

```python
def make_username(first_name, last_name):
    first = first_name.lower().replace(" ", "")
    last = last_name.lower().replace(" ", "")
    return first + "." + last
```

```python
make_username("Ada", "Lovelace")
```

```python
make_username("Alan", "Turing")
```

```python
make_username("Ana Maria", "De Leon")
```

## C. BOOKEND SWAP PROBLEM

```python
def swap_bookends(items):
    first, *middle, last = items
    return [last] + middle + [first]
```

```python
swap_bookends([1, 2, 3, 4, 5, 6])
```

```python
swap_bookends(["red", "green", "blue"])
```

```python
swap_bookends([8, 3])
```
